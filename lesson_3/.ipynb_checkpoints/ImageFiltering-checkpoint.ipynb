### Image Filtering
Image filtering transforms an image by combining each pixel with its neighbors, most commonly through **convolution**: sliding a small matrix (a *kernel*, or *filter*) across the image and computing a weighted sum at every position. Filtering is used to enhance visual quality, extract features (edges, corners, textures), and manipulate images — and since a convolutional layer is exactly this operation with learned weights, it is also the basic building block of convolutional neural networks.

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]
```

```python
img = cv2.imread('data/kodim01.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

### Convolution
To filter an image with a kernel, we place the kernel over each pixel, multiply every kernel weight by the pixel value underneath it, and sum the results — that sum becomes the output pixel. Let's implement this sliding-window operation by hand once, so it's clear what OpenCV's optimized functions are doing under the hood, before relying on those built-ins for everything else.

```python
kernel = np.array([[1, 1, 1], [1, 1, 1], [1, 1, 1]])/9
rows, cols, channels = img.shape
out = np.zeros_like(img)

# Sliding window applied to each colour channel.
# Border pixels (row/col 0 and the last row/col) are left at 0 since a 3x3
# window centred on them would go out of bounds without padding.
for r in range(1, rows - 1):
    for c in range(1, cols - 1):
        for ch in range(0, channels):
            block = img[r-1:r+2, c-1:c+2, ch]
            out[r, c, ch] = np.sum(block * kernel)
            
plt.imshow(out)
```

### High-pass Filters
OpenCV already includes a large set of predefined filters, and also lets us apply our own custom kernels through `cv2.filter2D`. High-pass filters amplify high-frequency content — sharp intensity changes such as edges and fine detail — while suppressing the smooth, low-frequency background. The kernel below is a simple **sharpening** filter: it boosts each pixel relative to its neighbors, making edges and texture more pronounced.

`ddepth` sets the pixel data type (bit depth) of the output — e.g. `CV_8U`, `CV_32F`. Passing `-1` means "same depth as the source image." This matters because a kernel with negative weights or a sum >1 (like this one) can push values outside 0–255; OpenCV saturates them to fit whatever `ddepth` you pick, so `-1` here just clips straight to 8-bit like the input.

```python
kernel = np.array([[-1, -1, -1], [-1, 9, -1], [-1, -1, -1]])
out = cv2.filter2D(img, ddepth=-1, kernel=kernel, borderType=cv2.BORDER_REPLICATE)
plt.imshow(out)
print(kernel)
```

### Low-pass Filters
Low-pass filters do the opposite of high-pass filters: they suppress high-frequency content and keep only the smooth, low-frequency part of the image, which is what makes them useful for blurring and denoising. The most common low-pass filter is the **Gaussian filter**, whose weights follow a Gaussian (bell-curve) distribution centered on the pixel being filtered, so nearby pixels contribute more than distant ones. OpenCV implements it efficiently with `cv2.GaussianBlur`, where `sigmaX` controls how wide that bell curve is — and therefore how strong the blur is.

#### Effect of sigma
`sigmaX` controls the spread of the Gaussian kernel and therefore how aggressively the image is blurred — a larger sigma averages over a wider neighborhood, trading detail for smoothness.

```python
sigmas = [1, 5, 15]
for i, sigma in enumerate(sigmas, start=1):
    # ksize=(0, 0) lets OpenCV derive an appropriately sized kernel from sigma
    blurred = cv2.GaussianBlur(img, ksize=(0, 0), sigmaX=sigma)
    plt.subplot(1, len(sigmas), i)
    plt.imshow(blurred)
    plt.title(f'sigma = {sigma}')
```

### Edge-preserving Filtering
Low-pass filters like the Gaussian blur remove noise, but they also blur every edge in the image, since they weight all neighboring pixels the same regardless of how similar they are to the center pixel. A **bilateral filter** fixes this by weighting neighbors by both their spatial distance (`sigmaSpace`) *and* how close their intensity is to the center pixel (`sigmaColor`). Pixels across an edge — where intensity changes sharply — end up contributing little, so edges are preserved while flat, noisy regions are still smoothed.

`d` sets the diameter (in pixels) of the neighborhood considered around each pixel — the bilateral filter's equivalent of kernel size in `GaussianBlur`/`filter2D`. It caps how far away a neighbor can be to influence the output pixel at all, independently of how `sigmaSpace`/`sigmaColor` weight pixels within that window. (If `d <= 0`, OpenCV derives the diameter from `sigmaSpace` instead.)

```python
# Or even simpler for predefined filters
out = cv2.GaussianBlur(img, ksize=(5,5), sigmaX=5)
plt.imshow(out)
```

```python
out = cv2.bilateralFilter(img, d=9, sigmaColor=75, sigmaSpace=75)
plt.imshow(out)
```

### Denoising
Denoising is the operation of removing (or reducing) noise from a signal. Let's corrupt an image with synthetic Gaussian noise and see how the low-pass and bilateral filters introduced above perform at removing it.

```python
img = cv2.imread('data/kodim07.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

noisy = img/255 + 0.1*np.random.randn(*img.shape)
noisy[noisy < 0] = 0
noisy[noisy > 1] = 1
noisy = (255*noisy).astype(np.uint8)

plt.subplot(121), plt.imshow(img), plt.title('Original')
plt.subplot(122), plt.imshow(noisy), plt.title('Noisy')
```

```python
# Apply Gaussian low-pass filter to reduce noise
out = cv2.GaussianBlur(noisy, ksize=(5,5), sigmaX=5)
plt.subplot(121), plt.imshow(noisy)
plt.subplot(122), plt.imshow(out)
```

Now let's compare the Gaussian-denoised result with the bilateral filter on the same noisy image, zoomed into a detailed region — the bilateral filter should keep edges noticeably sharper since it avoids blurring across them.

```python
plt.subplot(121), plt.imshow(noisy[0:300, 0:300, :])
plt.subplot(122), plt.imshow(out[0:300, 0:300, :])
```

```python
# Compare with bilateral filter
bilat = cv2.bilateralFilter(noisy, d=9, sigmaColor=75, sigmaSpace=75)
plt.subplot(121), plt.imshow(out[200:600, 300:600, :])
plt.subplot(122), plt.imshow(bilat[200:600, 300:600, :])
```

### Salt & Pepper Noise
Salt & pepper noise is impulsive: instead of a small random perturbation added to every pixel (like Gaussian noise), a random subset of pixels is corrupted outright, saturating to pure white ("salt") or pure black ("pepper"). We simulate it by marking a random fraction of pixels as corrupted, then setting each of those to white or black with equal probability.

```python
rows, cols, _ = img.shape
probability = 0.1

noisy = img.copy()
for r in range(rows):
    for c in range(cols):
        if np.random.rand() < probability:
            # this pixel gets corrupted: 50/50 chance of salt (white) or pepper (black)
            if np.random.rand() < 0.5:
                noisy[r, c, :] = 255
            else:
                noisy[r, c, :] = 0
        # otherwise noisy[r, c, :] already holds the original pixel, from img.copy()

plt.imshow(noisy)
```

Gaussian filtering worked well against additive Gaussian noise — let's see how it copes with impulsive noise instead.

```python
out = cv2.GaussianBlur(noisy, ksize=(5,5), sigmaX=5)
plt.subplot(121), plt.imshow(noisy)
plt.subplot(122), plt.imshow(out)
```

```python
plt.subplot(121), plt.imshow(noisy[200:600, 300:600, :])
plt.subplot(122), plt.imshow(out[200:600, 300:600, :])
```

### Median Filtering
Gaussian filters do not handle impulsive noise well: an averaging kernel still mixes the extreme salt/pepper values into their neighbors instead of removing them. A **median filter** instead replaces each pixel with the median value inside its neighborhood. Since a single outlier can't shift a median by much, impulsive noise is rejected almost entirely, while edges are preserved better than with an averaging (mean-based) filter.

```python
median = cv2.medianBlur(noisy, ksize=5)
plt.subplot(121), plt.imshow(noisy[200:600, 300:600, :])
plt.subplot(122), plt.imshow(median[200:600, 300:600, :])
```

### Takeaways
- **Convolution** with a kernel is the common mechanism behind sharpening, blurring, and (eventually) convolutional neural networks.
- **Low-pass filters** (Gaussian) remove noise but blur edges indiscriminately; **high-pass filters** sharpen edges but amplify noise along with them.
- **Bilateral filters** get edge-preserving smoothing by weighting neighbors by both spatial distance and intensity similarity.
- The best filter depends on the noise: Gaussian/bilateral filtering suits additive Gaussian noise, while **median filtering** is far better suited to impulsive salt & pepper noise.
